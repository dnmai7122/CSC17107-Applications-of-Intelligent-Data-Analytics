```python
duplicated_rows = data.duplicated().sum()

if duplicated_rows == 0:
    print('There are no duplicate rows, meaning each row in the DataFrame is unique.')
    print('So we do not need to keep processing duplicate lines')
else:
    print(f'There are {duplicated_rows} duplicate rows so we will delete {duplicated_rows} rows')
    data = data.drop_duplicates()
    print(f'After removing duplicate rows, there are {data.shape[0]} rows remaining')
```

   <span style="color:'black' ;">

   The `age`, `birthDay`, `birthMonth`, and `birthYear` columns in the dataset currently have incorrect data types, being stored as **float64** instead of the intended **int64**. Similarly, the `birthDate` and `date` columns are stored as **object**, whereas they should be of type **datetime64**. The next step involves converting these columns to their appropriate data types to ensure consistency and accuracy in the dataset.

   </span>
</div>

```python
cols_to_convert = ['age', 'birthDay', 'birthMonth', 'birthYear']
data[cols_to_convert] = data[cols_to_convert].fillna(0).astype('Int64')

date_cols = ['birthDate', 'date']
data[date_cols] = data[date_cols].apply(pd.to_datetime, errors='coerce')

data.dtypes.to_frame('Data Type')
```
